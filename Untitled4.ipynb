```python
def sobel_vertical_kernel_init(shape, dtype=None):
    # Sobel operator for vertical edge detection
    # The shape will be (kernel_height, kernel_width, input_channels, output_channels)
    # For greyscale, input_channels = 1, output_channels = 1
    kernel = np.array([[-1, 0, 1],
                       [-2, 0, 2],
                       [-1, 0, 1]], dtype=dtype)
    # Reshape the 2D kernel to 4D for Conv2D layer
    # (kernel_height, kernel_width, input_channels, output_channels)
    kernel = kernel.reshape(shape)
    return kernel

print("Sobel Vertical Kernel Example (for 3x3 kernel, 1 input channel, 1 output channel):")
print(sobel_vertical_kernel_init((3, 3, 1, 1), dtype=np.float32).squeeze())
```

Now, let's apply this kernel to the greyscale image using a `Conv2D` layer.

```python
conv_edge = Sequential([
    Conv2D(filters=1, kernel_size=(3, 3), padding="same",
           input_shape=(None, None, 1), # Greyscale image has 1 channel
           kernel_initializer=sobel_vertical_kernel_init,
           use_bias=False # Edge detection kernels usually don't use bias
          )
])

print(conv_edge.output_shape)
```

Finally, let's visualize the result of applying this edge detector to the greyscale image.

```python
img_in_grey = np.expand_dims(grey_sample_image, 0)

# Ensure the input is float32 as expected by Keras
img_in_grey_float = img_in_grey.astype(np.float32)

img_out_edge = conv_edge.predict(img_in_grey_float)

fig, (ax0, ax1) = plt.subplots(ncols=2, figsize=(12, 6))
ax0.imshow(np.squeeze(img_in_grey).astype(np.uint8),
           cmap=plt.cm.gray);
ax0.set_title('Original Greyscale Image')

# The output of Conv2D can have negative values, so we might need to normalize for display
# or clip values. For simplicity, let's normalize to 0-255.
img_out_display = img_out_edge[0].squeeze()
img_out_display = np.clip(img_out_display, 0, 255) # Clip values to be in display range

ax1.imshow(img_out_display.astype(np.uint8),
           cmap=plt.cm.gray);
ax1.set_title('Vertical Edge Detected Image')

plt.show()
```
